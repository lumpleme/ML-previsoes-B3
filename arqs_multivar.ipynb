```python
import json
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.graph_objects as go
```

## 1. Carregando dados para CSV

```python
with open('dados/petr4_multivar.json', 'r', encoding='utf-8') as f:
    arquivo_json = json.load(f)
```

### PETR4

```python
historico_petr4 = arquivo_json["data"]["history"]["tickers"][0]["history"]

df_petr4 = pd.DataFrame(historico_petr4)

# convertendo data e invertendo a ordem
df_petr4['date'] = pd.to_datetime(df_petr4['date'])
df_petr4 = df_petr4.sort_values(by='date', ascending=True).reset_index(drop=True)

# calculando retorno
df_petr4['return'] = df_petr4['closed'].pct_change()
# removendo o primeiro dia que ficou com NaN
df_petr4 = df_petr4.dropna().reset_index(drop=True)

# salvando arquivo
df_petr4 = df_petr4[['date', 'open', 'high', 'low', 'closed', 'vol', 'return']]
df_petr4.to_csv('dados/petr4_multivar.csv', index=False)
```
